```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All libraries imported successfully!")
```

```python
from io import StringIO

csv_data = """Name,Subject,Marks,Attendance
Asha,Math,85,95
Bala,Math,72,88
Charu,Science,90,92
Deepak,Science,,80
Esha,Math,65,75
Farhan,Science,78,
Gita,Math,95,98
Gita,Math,95,98
"""

df = pd.read_csv(StringIO(csv_data))

print("Dataset loaded successfully!")
display(df)
```

```python
print("First 5 rows:")
display(df.head())

print("\nDataset shape (rows, columns):")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())
```

```python
print("Missing values before cleaning:")
print(df.isnull().sum())

print("\nDuplicate rows before cleaning:")
print(df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates().copy()

# Fill missing marks with the median mark
df["Marks"] = df["Marks"].fillna(df["Marks"].median())

# Fill missing attendance with the median attendance
df["Attendance"] = df["Attendance"].fillna(
    df["Attendance"].median()
)

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nCleaned dataset:")
display(df)
```

```python
print("Total students:", len(df))
print("Average marks:", round(df["Marks"].mean(), 2))
print("Highest marks:", df["Marks"].max())
print("Lowest marks:", df["Marks"].min())
print("Average attendance:", round(df["Attendance"].mean(), 2))

# NumPy calculations
marks = df["Marks"].to_numpy()

print("\nNumPy Analysis")
print("Mean marks:", round(np.mean(marks), 2))
print("Median marks:", np.median(marks))
print("Standard deviation:", round(np.std(marks), 2))

# Average marks by subject
subject_average = df.groupby("Subject")["Marks"].mean()

print("\nAverage marks by subject:")
display(subject_average)
```

```python
subject_average.plot(
    kind="bar",
    title="Average Marks by Subject",
    xlabel="Subject",
    ylabel="Average Marks",
    rot=0
)

plt.tight_layout()
plt.show()
```

```python
plt.plot(
    df["Name"],
    df["Marks"],
    marker="o"
)

plt.title("Student Marks Comparison")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
subject_counts = df["Subject"].value_counts()

plt.pie(
    subject_counts.values,
    labels=subject_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Student Distribution by Subject")
plt.axis("equal")
plt.show()
```
